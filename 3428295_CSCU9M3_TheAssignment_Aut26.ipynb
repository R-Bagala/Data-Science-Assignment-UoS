## Exploratory Data Analysis
The dataset will now be analysed to understand what it contains and to observe any patterns, unusual features and to identify any interesting patterns. 

This section will carry out the analysis by answering 3 questions. The first question being the following: 

Does the number of recorded measurements change over the years?

The CSV file which will be used to answer this question is the clean_stats dataset as it contains several entries for a wide range of demographic statistics for each country and shows them beginning at 1960 and ending on the year 2021.

To properly visualize this, the total number of entries will need to be added for each year.

The following code will do so:

```python
#counting the number of entries for every year within the range
recorded_entries = clean_stats.loc[:, "1960":"2021"].count()

print(recorded_entries)
```

The results show that there is a noticeable increase in the number of statistical entries per year. The most noticeable gap being between 1960 and 2017 which is a large increase by 34354 and can be the result of data collection improving over time.

The number decreases rapidly to 40553 which is a decrease of 31551 and a possible reason being COVID disrupting some national surveys.

A graph will be used to show the gradual change and the matplotlib library will be installed for visualizations:

```python
#importing a library for visualizations
import matplotlib.pyplot as plt

#years will beused as an index
years = list(range(1960, 2022))

plt.plot(years, recorded_entries)

#making graph axis and titles
plt.xlabel("Year")
plt.ylabel("Number of entries")
plt.title("Availability of statistics over time.")

plt.show
```
